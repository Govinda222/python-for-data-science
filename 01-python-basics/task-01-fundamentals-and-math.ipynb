### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
import math
x1,y1 = map(int,input("Enter the coordinates of x1 and y1 ").split())
x2,y2 = map(int,input("Enter the coordinates of x2 and y2 ").split())
d = math.sqrt((x2-x1)**2 + (y2-y1)**2)
print (f"The distance between point a {x1,y1} and point b {x2,y2} is {d}")
````

````python
import matplotlib.pyplot as plt
#3. Plotting the visualization
plt.figure(figsize=(6, 6))

# Plot the two main points
plt.scatter([x1, x2], [y1, y2], color='red', zorder=5)
plt.text(x1, y1, f' A ({x1}, {y1})', verticalalignment='bottom', horizontalalignment='right', fontsize=12)
plt.text(x2, y2, f' B ({x2}, {y2})', verticalalignment='bottom', horizontalalignment='left', fontsize=12)

# Draw the direct Euclidean distance line (Hypotenuse)
plt.plot([x1, x2], [y1, y2], color='blue', linestyle='-', linewidth=2, label=f'Distance (d = {d:.2f})')

# Draw the dashed lines to show the horizontal and vertical changes (dx and dy)
plt.plot([x1, x2], [y1, y1], color='gray', linestyle='--', label=f'dx = {abs(x2-x1):.1f}')
plt.plot([x2, x2], [y1, y2], color='gray', linestyle='--', label=f'dy = {abs(y2-y1):.1f}')

# Graph aesthetics
plt.grid(True, which='both', linestyle=':', alpha=0.5)
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.title('Visualizing Euclidean Distance')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()

# Display the plot
plt.show()
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here

n = int(input("Enter the value of n: "))
total = 0
for i in range(1, n + 1):
    total += i ** 2
print("Sum of squares:", total)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here

a = int(input("Enter the first term: "))
second = int(input("Enter the second term: "))
n = int(input("Enter the value of n: "))
d = second - a
nth_term = a + (n - 1) * d
print("The Nth term is:", nth_term)
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# Write your code here

num1 = int(input("Enter numerator of first fraction: "))
den1 = int(input("Enter denominator of first fraction: "))

num2 = int(input("Enter numerator of second fraction: "))
den2 = int(input("Enter denominator of second fraction: "))

numerator = num1 * den2 + num2 * den1
denominator = den1 * den2

print("Sum of fractions:", numerator, "/", denominator)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here

H = float(input("Enter height of the milk tank (cm): "))
L = float(input("Enter length of the milk tank (cm): "))
B = float(input("Enter breadth of the milk tank (cm): "))

h = float(input("Enter height of the glass (cm): "))
r = float(input("Enter radius of the glass (cm): "))

tank_volume = H * L * B
glass_volume = (22/7) * r ** 2 * h

glasses = tank_volume // glass_volume

print("Number of glasses of milk:", int(glasses))
````
