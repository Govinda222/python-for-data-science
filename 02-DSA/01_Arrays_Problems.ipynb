```python
#Q1 find the kth smallest element in the list
def kth_sorted (arr,k):
    sorted_arr = sorted(arr)
    return sorted_arr[k-1]
arr = [7, 10, 4, 3, 20, 15]
k = 3
kth_sorted(arr,k)
```

```python
#Q2 check if the array is sorted
def is_sorted(arr):
    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            return False
    return True



# 🧪 The Gauntlet
test_cases = [
    ([1, 2, 3, 4, 5], True, "Standard sorted"),
    ([1, 3, 2, 4, 5], False, "Single inversion"),
    ([5, 4, 3, 2, 1], False, "Strictly descending"),
    ([42], True, "Single element"),
    ([], True, "Empty array"),
    ([2, 2, 2, 2], True, "All duplicates (non-decreasing)"),
    ([1, 2, 2, 3, 4], True, "Duplicates along the way"),
]

for arr, expected, label in test_cases:
    result = is_sorted(arr)
    status = "✅ PASS" if result == expected else f"❌ FAIL (Got {result}, Expected {expected})"
    print(f"[{status}] - {label}: {arr}")
```

```python
#Q3 find the min/max of the given array
arr = [12, 3, 45, -5, 22, 100, 0]
maximum = arr[0]
minimum = arr[0]
for i in arr:
    if i < minimum:
        minimum = i
    if i > maximum:
        maximum = i 
print(f"The minimum no is {minimum} and the maximum number is {maximum}")
```

```python
#Q4: First element to occur k times
arr = [2, 3, 4, 2, 2, 5, 3, 3]
k = 3

occ = {}
for i in arr:
    occ[i] = occ.get(i, 0) + 1

for i in arr:
    if occ[i] == k:
        print(i)
        break
```

```python
#Q5: Find duplicates in array
arr = [4, 3, 2, 7, 8, 2, 3, 1]
occ = {}
for i in arr:
    occ[i] = occ.get(i, 0) + 1

no_of_occ = [num for num, count in occ.items() if count >= 2]
print(no_of_occ)
```

```python
#Q10 Maximum sum subarray 
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
total = {}
for i in range (0,len(arr)):
    subarr = []
    for j in range(i,len(arr)):
        subarr.append(arr[j])
        total[sum(subarr)] = subarr
maximum = max(total.keys())
print(f"The maximum value is {maximum} and the sub array is {total[maximum]}")
```

```python
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

curr_sum = arr[0]
best_sum = arr[0]

curr_seq = [arr[0]]
best_seq = [arr[0]]

for i in arr[1:]:
    if curr_sum + i > i:
        curr_sum += i
        curr_seq.append(i)
    else:
        curr_sum = i
        curr_seq = [i]

    if curr_sum > best_sum:
        best_sum = curr_sum
        best_seq = curr_seq.copy()

    print(curr_seq,curr_sum)

print(f"Max Sum: {best_sum}")
print(f"Subarray: {best_seq}")
```
