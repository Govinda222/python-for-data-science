```python
# write a recursion to print multiplication of 2 digit without * 
def multiplication (a,b):
    if b == 1 :
        return a 
    else:
        return a+(multiplication(a,b-1))

a = int(input("Enter a number "))
b = int(input("Enter a number "))
print(multiplication(a,b))
```

```python
# factorial 
def fact (a):
    if a == 1:
        return 1
    else:
        return a*fact(a-1)
a = int(input("Enter a number "))
fact(a)
```

```python
# Palindrome string 
def palin(st):
    if len(st) <= 1:
        print("Palindrome")
    elif st[0] == st[-1]:
        palin(st[1:-1])
    else:
        print("Not a Palindrome")

palin("madam")  
palin("abba")  
```

```python
# recursive fibo 
def fibo (m):
    if m == 1 or m == 0:
        return 1 
    else :
        return fibo(m-1) + fibo(m-2)

fibo(10)
```
