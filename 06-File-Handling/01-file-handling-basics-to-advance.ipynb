# Introduction to with keyword 
## Using Context Manager (With)

* It's a good idea to close a file after usage as it will free up the resources
* If we dont close it, garbage collector would close it
* `with` keyword closes the file as soon as the usage is over

````python
# lets write using with 
with open("sample3.txt","w") as f:
    f.write("This is a demo file \nUsing with and it has no purpose \nJust kidding it is used to teach me file handling")
````

````python
f.read()# cant open as with auto collects garbage 
````

````python
# now lets read via with 
with open("sample3.txt","r") as f:
    print(f.read())
````

````python
# understanding buffer here python has a buffer which keep tracks of how many char is printed 
# that's why we are printing the unprinted line and using 
# this we can load even a 10gb file in a 2 gb roam 
with open("sample3.txt","r") as f:
    print(f.read(10))
    print(f.read(10))
````

````python
# lets understand how 
Big_text = ["Hello World\n" for i in range (1000)] # created a big file now let's read them in chunks 
with open("Big_file.txt","w") as f:
    f.writelines(Big_text)
````

````python
with open("Big_file.txt", "r") as f:
    chunk_size = 100
    while chunk := f.read(chunk_size):
        print(chunk, end="")
````

````python
# seek and tell 
# tell tells us what is the next char to be printed 
# seek is used to move the buffer in any of the char 
# example 
with open("sample3.txt","r") as f:
    print(f.read(10))

    print(f.tell()) # it gives 10 as buffer starts with 0 and goes till n-1 so here 10 means 11 
````

````python
with open("sample3.txt","r") as f:
    print(f.read(10))
    f.seek(2) # here i reset the position of buffer to 2nd index char as its n-1 
    print(f.read(10))
````

## Serialization and Deserialization

* **Serialization** - process of converting python data types to JSON format
* **Deserialization** - process of converting JSON to python data types

What is JSON?

```json
{
    "d": {
        "results": [
            {
                "__metadata": {
                    "type": "EmployeeDetails.Employee"
                },
                "UserID": "E12012",
                "RoleCode": "35"
            }
        ]
    }
}

````python
#serialization using list
import json
l = [1,2,3,4] 
with open("demo.json","w") as f:
    json.dump(l,f) # in dump u pass 2 things the value to be added and the file where it should be added 
````

````python
#serialization using dict
d = {
    "name":"Govinda",
    "Age" : 23,
    "Gender" : "Male" 
     }
import json
with open("demo.json","w") as f:
    json.dump(d,f,indent=4) 
````

````python
# now lets try deserialization 
import json 
with open("demo.json", "r") as f:
    d = json.load(f) # here we use load and dump instead of read and write 
    print(d)
    print(type(d)) # best part the data type remains the same 
````

````python
# serialization and deserialization using nested dict 
d = {
    "name":"Govinda",
    "Age" : 23,
    "Gender" : "Male",
    "marks" : {
        "english":30,
        "maths":50,
        "computer":100
        }
     }
import json
with open("demo.json","w") as f:
    json.dump(d,f,indent=4) 
````

````python
import json 
with open("demo.json", "r") as f:
    d = json.load(f)
    print(d)
````

````python
# serialization and deserialization of custom class 
import json

class Person:
    def __init__(self, fname, lname, age, gender):
        self.fname = fname
        self.lname = lname
        self.age = age 
        self.gender = gender

p = Person("Govinda", "Upadhyay", 23, "Male")        

def show(obj):
    if isinstance(obj, Person):
        return f"{obj.fname} {obj.lname} -> {obj.age} -> {obj.gender}"

with open("demo2.json", "w") as f:
    json.dump(p, f, default=show) 
````

````python
# we can also do it in dict via 
import json

class Person:
    def __init__(self, fname, lname, age, gender):
        self.fname = fname
        self.lname = lname
        self.age = age 
        self.gender = gender

p = Person("Govinda", "Upadhyay", 23, "Male")        

def show(obj):
    if isinstance(obj, Person):
        return {
            "full_name": f"{obj.fname} {obj.lname}",
            "age": obj.age,
            "gender": obj.gender
        }

with open("demo3.json", "w") as f:
    json.dump(p, f, default=show, indent=4)
````

## Pickling

`Pickling` is the process whereby a Python object hierarchy is converted into a byte stream, and `unpickling` is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

````python
class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is',self.name,'and I am ',self.age,'years old')
````

````python
p = Person ("Govinda",23)
````

````python
# last method Pickle 
import pickle
with open("Person.pkl","wb") as f :
    pickle.dump(p,f)
````

````python
# this is cool as it can also use its attributes 
import pickle
with open("Person.pkl","rb") as f :
    p = pickle.load(f)
p.display_info()
````

## Pickle Vs Json

* Pickle lets the user to store data in binary format. JSON lets the user store data in a human-readable text format.[cite: 6]
* **Language Support:** Pickle is Python-specific, whereas JSON is language-independent and supported across almost all programming languages.
* **Data Types:** Pickle can serialize almost any Python object (including custom classes, functions, and tuples), while JSON only supports basic data types (strings, numbers, booleans, lists/arrays, and dictionaries/objects).
* **Security:** Unpickling data from untrusted sources is insecure as it can execute arbitrary code, whereas JSON is safe to parse since it only stores data.
* **File Size & Speed:** Pickle is generally faster to serialize/deserialize and produces more compact files for complex Python objects compared to JSON.
