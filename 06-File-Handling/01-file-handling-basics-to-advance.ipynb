```python
#Q1 create a file and write on it 
f=open("sample.txt","w") # w here is write mode and if the file doesnot exist we create it using that 
f.write("Hello World")
f.close() 
```

```python
# create a file and write multi line 
f = open ("sample2.txt","w")
f.write("Hello \nHow are u doing \nI am fine what about you ? ")
f.close()
```

```python
# how to write an already created file 
f = open("sample.txt","w")
f.write("Well its a demo on how to re write a file that's all")
f.close()
```
