# Problem-2: Student Management System in Python

## Objective
Build a menu-driven **Student Management System** in Python (using OOP) that stores student records (`Name`, `RollNo`, `Marks1`, `Marks2`) and supports the following 6 operations[cite: 5, 6]:

---

## Required Operations
1. **Accept (`accept`)** — Takes student details from the user: `Name`, `RollNo`, and marks for two subjects (`Marks1`, `Marks2`).
2. **Display (`display`)** — Displays the details of every stored student[cite: 5].
3. **Search (`search`)** — Asks the user for a `RollNo`, finds the matching student, and prints their details[cite: 5].
4. **Delete (`delete`)** — Asks the user for a `RollNo` and deletes the matching student's record[cite: 5].
5. **Update (`update`)** — Asks for an `old_roll_no` and a `new_roll_no`, and updates that student's roll number[cite: 5].
6. **Exit** — Exits the program[cite: 6].

---

## Expected Output Format

### List of Students (Display)
```text
Name   : A
RollNo : 1
Marks1 : 100
Marks2 : 100

Name   : B
RollNo : 2
Marks1 : 90
Marks2 : 90

Name   : C
RollNo : 3
Marks1 : 80
Marks2 : 80

````python
class Student:
    def __init__(self):
        self.records = {}  
        self.menu()

    def menu(self):
        choice = input("""
Enter 1 to enter a record
Enter 2 to display all the records
Enter 3 to search a student
Enter 4 to delete a student
Enter 5 to update a student
Enter 6 to exit
Enter your choice: """).strip()

        if choice == "1":
            self.accept()
        elif choice == "2":
            self.display()
        elif choice == "3":
            self.search()
        elif choice == "4":
            self.delete()
        elif choice == "5":
            self.update()
        elif choice == "6":
            print("Exiting Student Management System.")
            return
        else:
            print("Invalid choice")
            self.menu()

    def accept(self):
        name = input("Enter your name")
        rollno = int(input("Enter your rollno"))
        marks1 = int(input("Enter your first subject marks"))
        marks2 = int(input("Enter your second subject marks"))
        if rollno in self.records:
            print("Rollno already exists kindly update instead")
        else:
            self.records[rollno] = [name,marks1,marks2]
        self.menu()


    def search(self):
        rollno = int(input("Enter the rollno you are looking for: "))
        if rollno in self.records:
            print("Student Found,")
            print(f"Name   : {self.records[rollno][0]}")
            print(f"RollNo : {rollno}")
            print(f"Marks1 : {self.records[rollno][1]}")
            print(f"Marks2 : {self.records[rollno][2]}")
        else:
            print("Student not found, please try again")
        self.menu()


    def display(self):
        if len(self.records) > 0:
            print("The records are as Follows : ")
            for i in self.records.keys():
                print(f"Name   : {self.records[i][0]}")
                print(f"RollNo : {i}")
                print(f"Marks1 : {self.records[i][1]}")
                print(f"Marks2 : {self.records[i][2]}")
        else:
            print("There is no records in the Database")
        self.menu()


    def delete(self):
        rollno = int(input("Enter the roll no you want to delete : "))
        if rollno in self.records:
            del(self.records[rollno])
            print("Rollno Deletion was successful")
        else:
            print("Roll no not found please try deleting someone else ")
        self.menu()

    def update(self):
            rollno = int(input("Enter the roll no you want to update: "))
            if rollno in self.records:
                sub_choice = input("""
    What would you like to update?
    1. Roll Number
    2. Name
    3. Marks1
    4. Marks2
    5. All Details
    Enter your choice: """).strip()

                if sub_choice == "1":
                    new_rollno = int(input("Enter the new roll no: "))
                    if new_rollno != rollno and new_rollno in self.records:
                        print("New roll no already exists, try a different one.")
                    else:
                        self.records[new_rollno] = self.records.pop(rollno)
                        print("Roll number updated successfully!")

                elif sub_choice == "2":
                    new_name = input("Enter the updated name: ")
                    self.records[rollno][0] = new_name
                    print("Name updated successfully!")

                elif sub_choice == "3":
                    new_marks1 = int(input("Enter the updated Marks1: "))
                    self.records[rollno][1] = new_marks1
                    print("Marks1 updated successfully!")

                elif sub_choice == "4":
                    new_marks2 = int(input("Enter the updated Marks2: "))
                    self.records[rollno][2] = new_marks2
                    print("Marks2 updated successfully!")

                elif sub_choice == "5":
                    new_rollno = int(input("Enter the new roll no: "))
                    if new_rollno != rollno and new_rollno in self.records:
                        print("New roll no already exists, update cancelled.")
                    else:
                        new_name = input("Enter the updated name: ")
                        new_marks1 = int(input("Enter the updated Marks1: "))
                        new_marks2 = int(input("Enter the updated Marks2: "))
                        self.records.pop(rollno)
                        self.records[new_rollno] = [new_name, new_marks1, new_marks2]
                        print("All records updated successfully!")

                else:
                    print("Invalid choice.")
            else:
                print("Rollno does not exist, kindly add it.")

            self.menu()
````
