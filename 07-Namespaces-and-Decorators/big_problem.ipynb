```python
def sanity_check(data_type):
    def outer(func):
        def inner(*args):
            for arg in args:
                if type(arg) != data_type:
                    raise TypeError("Wrong Type of data")
            return func(*args)
        return inner
    return outer
```

```python
@sanity_check(int)
def add(a, b):
    return a + b
print(add(3, 5))  
print(add(3, "5"))
```
