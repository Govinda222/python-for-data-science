# Write the OOP class to handle the following 
1. User can create and view 2D cordinates 
2. User can find the distance between 2 coordinate 
3. User can find the distance of a cordinate from the origin
4. User can check if a point lies on a given line 
5. User can find distance between a given 2D point and a given line 

```python
import math
class Point :
    def __init__(self,x,y):
        self.x_corr = x
        self.y_corr = y

    def __str__(self):
        return "<{},{}>".format(self.x_corr,self.y_corr)
    
    def euclidean_distance (self,other):
        return ((self.x_corr-other.x_corr)**2 + (self.y_corr-other.y_corr)**2)**0.5 

    def distance_from_origin (self):
        return(self.euclidean_distance(Point(0,0)))

class Line:
    def __init__(self,A,B,C):
        self.A = A 
        self.B = B
        self.C = C 

    def __str__(self):
        return f"{self.A}x {self.B:+}y {self.C:+} = 0"

    def is_point_on_line(self, point):
        val = (self.A * point.x_corr) + (self.B * point.y_corr) + self.C
        return math.isclose(val, 0, abs_tol=1e-9)
    
    def distance_between_point_and_line(self,point):
        return abs((self.A*point.x_corr) + (self.B*point.y_corr)+self.C)/math.sqrt(self.A**2+self.B**2)

    def intersection(self, other):
        denom = (self.A * other.B) - (other.A * self.B)
    
        if denom == 0:
            return "Lines are parallel or identical (no unique intersection)"
        
        new_x = ((self.B * other.C) - (other.B * self.C)) / denom
        new_y = ((self.C * other.A) - (other.C * self.A)) / denom
    
        return Point(new_x, new_y)
```

```python
p1 = Point(2, 3)
p2 = Point(5, 7)

print("Point 1:", p1)                                      # <2,3>
print("Distance p1 to p2:", p1.euclidean_distance(p2))      # 5.0
print("Distance from origin:", p2.distance_from_origin())  # 8.602325267042627

# Line 1: x - y + 1 = 0  (passes through <2,3>)
l1 = Line(1, -1, 1)
# Line 2: x + y - 5 = 0  (intersects l1 at <2,3>)
l2 = Line(1, 1, -5)

print("Line 1:", l1)                                       # 1x -1y +1 = 0
print("Is p1 on l1?:", l1.is_point_on_line(p1))            # True
print("Distance from p2 to l1:", l1.distance_between_point_and_line(p2))  # 0.7071067811865475
print("Intersection of l1 & l2:", l1.intersection(l2))     # <2.0,3.0>
```

# Static Variables (Vs Instance Variable)

1. Static variable remains static means it doesnot change eg name of the clg and all 
2. Instance variable is something that can be changed for all obj eg cgpa of the student 
3. in the already build code we want to implement a static var to add a cus id like 101,102 and so on 
4. Static variable is named after the class so its class.var name where as instance is named after obj or self.var name

```python
class ATM:
    __counter = 100 # this is a static variable
    def __init__(self):
        self.balance = 0
        self.pin = ""
        self.menu()
        self.cus_id = ATM.__counter # this is how we access it __ is to make pvt var 
        ATM.__counter += 1 # updating the var 

    #now lets make getter
    @staticmethod
    def get_counter():
        return ATM.__counter

    # now lets make setter
    @staticmethod
    def set_counter(value):
        ATM.__counter = value

    def menu(self):
        while True:
            user_input = input("""
1. Create PIN
2. Change PIN
3. Check Balance
4. Withdraw
5. Deposit
6. Exit
Choose an option: """)
            if user_input == "1":
                self.create_pin()
            elif user_input == "2":
                self.change_pin()
            elif user_input == "3":
                self.check_balance()
            elif user_input == "4":
                self.withdraw()
            elif user_input == "5":
                self.deposit()
            elif user_input == "6":
                print("Thank you for using the ATM.")
                break
            else:
                print("Invalid choice. Please enter 1-6.")

    def _verify_pin(self):
        if not self.pin:
            print("No PIN set. Please create a PIN first.")
            return False
        user_pin = input("Enter your PIN: ")
        if user_pin != self.pin:
            print("Invalid PIN.")
            return False
        return True

    def create_pin(self):
        user_pin = input("Enter the new PIN (min 4 digits): ")
        confirm = input("Confirm your PIN: ")
        if user_pin != confirm or len(user_pin) < 4 or not user_pin.isdigit():
            print("Invalid PIN. Must be at least 4 digits and match.")
        else:
            self.pin = user_pin
            print("PIN created successfully.")

    def change_pin(self):
        if self._verify_pin():
            new_pin = input("Enter the new PIN (min 4 digits): ")
            confirm = input("Confirm new PIN: ")
            if new_pin != confirm or len(new_pin) < 4 or not new_pin.isdigit():
                print("Invalid new PIN.")
            else:
                self.pin = new_pin
                print("PIN changed successfully.")

    def check_balance(self):
        if self._verify_pin():
            print(f"Your available balance is: {self.balance}")

    def withdraw(self):
        if self._verify_pin():
            try:
                amt = int(input("Enter amount to withdraw: "))
                if amt <= 0 or amt > self.balance:
                    print("Invalid withdrawal amount or insufficient balance.")
                else:
                    self.balance -= amt
                    print(f"Withdrawal successful. Remaining balance: {self.balance}")
            except ValueError:
                print("Please enter a valid numeric amount.")

    def deposit(self):
        if self._verify_pin():
            try:
                amt = int(input("Enter amount to deposit: "))
                if amt <= 0:
                    print("Deposit amount must be greater than 0.")
                else:
                    self.balance += amt
                    print(f"Deposit successful. Updated balance: {self.balance}")
            except ValueError:
                print("Please enter a valid numeric amount.")
```
