Note: The bus seating capacity is 50. so the final fare amount should be 5500. You need to override the fare() method of a Vehicle class in Bus class.

````python
# Write code here
class Vehicle ():
    def __init__(self,type,capacity):
        self.type = type 
        self.capacity = capacity

    def fare (self):
        return self.capacity * 100

class Bus(Vehicle):
    def fare(self):
        total_fare = super().fare()
        return total_fare + total_fare * 0.1


Bus1 = Bus("bus",50)
Bus1.fare()
````

### `Problem-3:` Write a program that has a class Point. Define another class Location which has two objects (Location & Destination) of class Point. Also define a function in Location that prints the reflection of Destination on the x axis.

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"{self.x}:{self.y}"


class Location:
    def __init__(self, location, destination):
        self.location = location
        self.destination = destination

    def reflection(self):
        self.destination.y = -self.destination.y
        print(f"The reflection on x axis is {self.destination}")


loc = Point(2, 3)
dest = Point(4, 5)

trip = Location(loc, dest)
trip.reflection()
````

### `Problem-4:` Write a program that has an abstract class Polygon. Derive two classes Rectangle and Triamgle from Polygon and write methods to get the details of their dimensions and hence calculate the area.

````python
from abc import ABC,abstractmethod
class Polygon(ABC):
    @abstractmethod
    def getdata(self):
        pass
    @abstractmethod
    def area(self):
        pass
    
class Rectangle (Polygon):
    def getdata(self,l,b):
        self.l = l
        self.b = b 

    def area(self):
        return self.l*self.b

class Triangle (Polygon):
    def getdata(self,h,b):
        self.h = h
        self.b = b 

    def area(self):
        return 0.5*(self.h*self.b)

rect = Rectangle()
rect.getdata(10, 5)
print("Rectangle Area:", rect.area()) 

tri = Triangle()
tri.getdata(8, 6)
print("Triangle Area:", tri.area()) 
````

### `Problem-5:` Write a program with class Bill. The users have the option to pay the bill either by cheque or by cash. Use the inheritance to model this situation.

````python
# Base Class
class Bill:
    def __init__(self, customer_name, amount):
        self.customer_name = customer_name
        self.amount = amount

    def display_bill(self):
        print("\n--- BILL RECEIPT ---")
        print(f"Customer Name : {self.customer_name}")
        print(f"Total Amount  : ₹{self.amount:.2f}")


class ChequePayment(Bill):
    def __init__(self, customer_name, amount, cheque_number, bank_name):
        super().__init__(customer_name, amount)
        self.cheque_number = cheque_number
        self.bank_name = bank_name

    def display_receipt(self):
        self.display_bill()
        print("Payment Mode  : Cheque")
        print(f"Cheque Number : {self.cheque_number}")
        print(f"Bank Name     : {self.bank_name}")
        print("Status        : Subject to realisation\n")

class CashPayment(Bill):
    def __init__(self, customer_name, amount, cash_tendered):
        super().__init__(customer_name, amount)
        self.cash_tendered = cash_tendered
        self.change = cash_tendered - amount

    def display_receipt(self):
        self.display_bill()
        print("Payment Mode  : Cash")
        print(f"Cash Tendered : ₹{self.cash_tendered:.2f}")
        if self.change >= 0:
            print(f"Change Return : ₹{self.change:.2f}")
            print("Status        : Paid in full\n")
        else:
            print(f"Balance Due   : ₹{abs(self.change):.2f}")
            print("Status        : Partial payment\n")


def main():
    name = input("Enter Customer Name: ")
    amount = float(input("Enter Bill Amount (₹): "))
    
    print("\nSelect Payment Method:")
    print("1. Cheque")
    print("2. Cash")
    choice = input("Enter choice (1 or 2): ")

    if choice == '1':
        cheque_no = input("Enter Cheque Number: ")
        bank = input("Enter Bank Name: ")
        payment = ChequePayment(name, amount, cheque_no, bank)
        payment.display_receipt()

    elif choice == '2':
        cash = float(input("Enter Cash Tendered (₹): "))
        payment = CashPayment(name, amount, cash)
        payment.display_receipt()

    else:
        print("Invalid payment option selected.")

p1 = ChequePayment("Govinda", 2500, "CHQ987654", "SBI")
p1.display_receipt()

p2 = CashPayment("Ankit", 1350.50, 1500)
p2.display_receipt()

p3 = CashPayment("Rohan", 800, 500)
p3.display_receipt()
````

###`Q-6:` FlexibleDict
As of now we are accessing values from dictionary with exact keys. Now we want to amend accessing values functionality. if a dict have key `1` (int) the even if we try to access values by giving `'1'` (1 as str) as key, we should get the same result and vice versa.

Write a class `FlexibleDict` upon builtin `dict` class with above required functionality.

Hint- `dict[key] => dict.__getitem__(key)`

Ex.
```
fd = FlexibleDict()
fd['a'] = 100
print(fd['a']) # Like regular dict

fd[5] = 500
print(fd[5]) # Like regular dict

fd[1] = 100
print(fd['1']) # actual Key is int but still trying to access through str key.
fd['1'] = 100
print(fd[1])

```
`Output:`
```
100
500
100
100

```

````python
class FlexibleDict(dict):
    def __getitem__(self, key):
        try:
            return super().__getitem__(key)
        except KeyError:
            pass

        if isinstance(key, int) and not isinstance(key, bool):
            alt_key = str(key)
            if super().__contains__(alt_key):
                return super().__getitem__(alt_key)

        elif isinstance(key, str):
            try:
                alt_key = int(key)
                if super().__contains__(alt_key):
                    return super().__getitem__(alt_key)
            except ValueError:
                pass

        # 4. Raise standard KeyError if neither form exists
        raise KeyError(key)


fd = FlexibleDict({1: 'One', '2': 'Two', 'name': 'Python'})

print(fd[1])       
print(fd['1'])     

print(fd['2'])    
print(fd[2])       
print(fd['name'])
````
