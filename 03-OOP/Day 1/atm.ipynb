```python
class ATM:
    def __init__(self):
        self.balance = 0
        self.pin = ""
        self.menu()

    def menu(self):
        while True:
            user_input = input("""
1. Create PIN
2. Change PIN
3. Check Balance
4. Withdraw
5. Deposit
6. Exit
Choose an option: """)
            if user_input == "1":
                self.create_pin()
            elif user_input == "2":
                self.change_pin()
            elif user_input == "3":
                self.check_balance()
            elif user_input == "4":
                self.withdraw()
            elif user_input == "5":
                self.deposit()
            elif user_input == "6":
                print("Thank you for using the ATM.")
                break
            else:
                print("Invalid choice. Please enter 1-6.")

    def _verify_pin(self):
        if not self.pin:
            print("No PIN set. Please create a PIN first.")
            return False
        user_pin = input("Enter your PIN: ")
        if user_pin != self.pin:
            print("Invalid PIN.")
            return False
        return True

    def create_pin(self):
        user_pin = input("Enter the new PIN (min 4 digits): ")
        confirm = input("Confirm your PIN: ")
        if user_pin != confirm or len(user_pin) < 4 or not user_pin.isdigit():
            print("Invalid PIN. Must be at least 4 digits and match.")
        else:
            self.pin = user_pin
            print("PIN created successfully.")

    def change_pin(self):
        if self._verify_pin():
            new_pin = input("Enter the new PIN (min 4 digits): ")
            confirm = input("Confirm new PIN: ")
            if new_pin != confirm or len(new_pin) < 4 or not new_pin.isdigit():
                print("Invalid new PIN.")
            else:
                self.pin = new_pin
                print("PIN changed successfully.")

    def check_balance(self):
        if self._verify_pin():
            print(f"Your available balance is: {self.balance}")

    def withdraw(self):
        if self._verify_pin():
            try:
                amt = int(input("Enter amount to withdraw: "))
                if amt <= 0 or amt > self.balance:
                    print("Invalid withdrawal amount or insufficient balance.")
                else:
                    self.balance -= amt
                    print(f"Withdrawal successful. Remaining balance: {self.balance}")
            except ValueError:
                print("Please enter a valid numeric amount.")

    def deposit(self):
        if self._verify_pin():
            try:
                amt = int(input("Enter amount to deposit: "))
                if amt <= 0:
                    print("Deposit amount must be greater than 0.")
                else:
                    self.balance += amt
                    print(f"Deposit successful. Updated balance: {self.balance}")
            except ValueError:
                print("Please enter a valid numeric amount.")
```
