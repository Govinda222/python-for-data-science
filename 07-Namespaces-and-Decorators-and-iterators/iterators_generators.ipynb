### What is Iteration? 

Iteration is a general term for taking each term of something one after another, anytime you use loops on either implicit or explicit to go over the group of items its called as iteration

```python
#Example
num = [1,2,3]
for i in num:
    print(i)
```

### What is Iterator?
Iteration is an object which allows the programmer to traverse through a sequence of data without having them to store the entire data in the memory
* iterator is the reason why iteration happens

```python
import sys

# Eager: builds the entire list in memory
L = [i for i in range(1, 100000)]
print(f"The size of L in KB is : {sys.getsizeof(L) / 1024:.2f} KB")

# Lazy: only stores start, stop, and step
x = range(1, 100000)
print(f"The size of x in KB is : {sys.getsizeof(x) / 1024:.4f} KB")
```

# we can make our own loop using this 

```python
def personal_loop(iterable):
    iterator = iter(iterable)
    while True:
        try:
            print(next(iterator))
        except StopIteration:
            break
```

```python
a = [1,2,3]
b = range(1,11)
c = (1,2,3)
d = {1,2,3}
e = {0:1,1:2}
```

```python
# making the personal range
class personal_range:
    def __init__(self, start, end):
        self.start = start
        self.end = end

    def __iter__(self):
        return personal_range_iterator(self)


class personal_range_iterator:
    def __init__(self, iterable_obj):
        self.iterable = iterable_obj
        self.current = iterable_obj.start 

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= self.iterable.end:
            raise StopIteration
        
        value = self.current
        self.current += 1
        return value
```

```python
personal_loop(personal_range(1,11))
```

```python
r = personal_range(1, 4)
print(list(r))
```

```python
r = personal_range(1, 4)
it = iter(r)
print(list(it)) # Line 3
print(list(it)) # Line 4
```

# What is generator ? 

its the simple way to make iterators

```python
# personal range using generators 

def personal_range(start, end):
    current = start
    while current < end:
        yield current
        current += 1
```

```python
# personal loop using generators 

def personal_loop(iterable):
    iterator = iter(iterable)
    while True:
        try :
            print(next(iterator))
        except StopIteration:
            break
```

```python
personal_loop(personal_range(1,6))
```

### Why is a Generator Expression Better Than a List Comprehension?

A **generator expression** (`(...)`) is better than a **list comprehension** (`[...]`) when working with large datasets or data streams because it uses **lazy evaluation**—computing and yielding one item at a time instead of building and storing the entire collection in memory upfront.

| Feature | List Comprehension (`[...]`) | Generator Expression (`(...)`) |
| :--- | :--- | :--- |
| **Memory Usage** | **High ($O(N)$):** Allocates RAM for all items immediately. | **Constant ($O(1)$):** Uses ~100–200 bytes regardless of dataset size. |
| **Execution** | **Eager:** Computes every element before returning. | **Lazy:** Computes the next element only when requested via `next()`. |
| **Short-Circuiting** | Computes everything even if you only need the first match. | Stops immediately once a condition is met (e.g., `next()`, `any()`). |
| **Reusability** | **Reusable:** Supports indexing (`L[0]`), `len()`, and multiple loops. | **Single-use:** Exhausts after one pass; no indexing or `len()`. |

#### Example

```python
import sys

list_comp = [i**2 for i in range(1_000_000)]

gen_exp = (i**2 for i in range(1_000_000))

print(sys.getsizeof(list_comp))  # ~8,448,728 bytes (~8.4 MB)
print(sys.getsizeof(gen_exp))    # ~200 bytes
```
